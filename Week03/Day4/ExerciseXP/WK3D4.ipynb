Using the code above, implement the relevant methods and dunder methods which will output the results below.

Hint : When adding 2 currencies which don’t share the same label you should raise an error.


```python
class Currency:
    def __init__(self, currency, amount):
        self.currency = currency
        self.amount = amount

    def __str__(self):
      return f"{self.amount} {self.currency}s"

    def __repr__(self):
      return f"{self.amount} {self.currency}s"


    def __int__(self):
       return self.amount

    def __add__(self, other):
      if isinstance(other, int):
        return self.amount + other
      elif isinstance(other, Currency):
        if self.currency==other.currency:
         return self.amount + other.amount
        else:
          raise TypeError(f"Cannot add between Currency type <{self.currency}> and <{other.currency}>")

    def __iadd__(self, other):
      if isinstance(other, int):
          self.amount += other
      elif isinstance(other, Currency):
          if self.currency == other.currency:
              self.amount += other.amount
          else:
              raise TypeError(f"Cannot add between Currency type <{self.currency}> and <{other.currency}>")
      else:
          raise TypeError("Unsupported type for += operation")
      return self



c1 = Currency('dollar', 5)
c2 = Currency('dollar', 10)
c3 = Currency('shekel', 1)
c4 = Currency('shekel', 10)


print(str(c1))
print(int(c1))
print(repr(c1))
print(c1)
c1 += 5
print(c1)

c1 += c2
print(c1)

c1 + c3
print(c1)
```

Exercise 5 : Amount of time left until January 1st


---


Create a function that displays the amount of time left from now until January 1st.

(Example: the 1st of January is in 10 days and 10:34:01hours).


```python
import datetime
from datetime import date


def nextYear():
  now = datetime.datetime.now()
  today= now.date()
  NY26 = date(2026, 1, 1)
  days_until = (NY26 - today)
  print(f" There are {days_until.days} days until the 1st of January 2026.")

nextYear()

```

Create a function that accepts a birthdate as an argument (in the format of your choice), then displays a message stating how many minutes the user lived in his life.


```python
import datetime
from datetime import date

def userBirthday():
    name = input ("What's your name? ") # not part of the exercise, but thought I would add this :)
    bDay_q = input("What's your date of birth? ")
    try:
        bDay=datetime.datetime.strptime(bDay_q,"%Y-%m-%d")
    except ValueError:
      print("Incorrect birthdate format, should be YYYY-MM-dd")
      return
    now= datetime.datetime.now()
    today= now

    dayAge = (today-bDay)
    minAge = int(dayAge.total_seconds()/60) # we have to convert to seconds and then to minutes by using int to truncate to the minute
    print(f" Hi {name}, you have been alive for {minAge: ,} minutes in your life!")
userBirthday()
```

Use faker to populate them with fake data.

```python
!pip install faker
from faker import Faker
import pandas as pd
fake = Faker(locale= 'en_us')# specify the language/location
```

```python

def createUsers(numUsers):
  users = []
  for i in range(numUsers):
    person = {}
    person['name']=fake.name()
    person['address']=fake.address()
    person['language_code']=fake.language_code()
    users.append(person)
  return users

createUsers(10)
```
