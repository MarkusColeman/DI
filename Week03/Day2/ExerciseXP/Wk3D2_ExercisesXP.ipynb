1. Create a class called Dog with the following attributes name, age, weight.
2. Implement the following methods in the Dog class:
    * bark: returns a string which states: “{dog_name} is barking”.
    * run_speed: returns the dogs running speed (weight/age*10).
    * fight : takes a parameter which value is another Dog instance, called other_dog. This method returns a string stating which dog won the fight. The winner should be the dog with the higher run_speed x weight.
3. Create 3 dogs and run them through your class.

```python
class Dog ():
  def __init__(self, name, age, weight):
    self.name = name
    self.age = age
    self.weight = weight

  def bark(self):
    return f'{self.name} is barking'

  def run_speed(self):
    return self.weight / self.age * 10

  def fight(self, other_dog):
    d1 = self.run_speed() * self.weight
    d2 = other_dog.run_speed() * other_dog.weight
    if d1 > d2:
      return f'{self.name} wins the battle!'
    else:
      return f'{other_dog.name} wins the battle!'

dog1 = Dog("Blaze", 8, 12)
dog2 = Dog("Shadow", 10, 9)
dog3 = Dog("Milo", 3, 20)

print(dog1.fight(dog2))
print(dog1.fight(dog3))
print(dog2.fight(dog3))



```

```python
#this simulates new file in Colab
%%writefile dog.py

class Dog:
    def __init__(self, name, age, weight):
        self.name = name
        self.age = age
        self.weight = weight

    def bark(self):
        return f'{self.name} is barking'

    def run_speed(self):
        return self.weight / self.age * 10

    def fight(self, other_dog):
        d1 = self.run_speed() * self.weight
        d2 = other_dog.run_speed() * other_dog.weight
        if d1 > d2:
            return f'{self.name} wins the battle!'
        else:
            return f'{other_dog.name} wins the battle!'
```

```python
from dog import Dog
import random



class PetDog(Dog):
  def __init__(self, name, age, weight, trained = False):
    super().__init__(name, age, weight)
    self.trained = trained


  def train(self, trained = True):
    self.trained = True
    print(self.bark())

  def play(self, *args):
    all_names=[self.name] + [dog.name for dog in args]
    names_string = ", ".join(all_names)
    print(f'{names_string} all play together')


  def do_a_trick(self):
    tricks = ["rolls over", "plays dead", "shakes your hand"]
    if self.trained == True:
      print(f'{self.name} {random.choice(tricks)}')


dog1 = PetDog("Jack", 2, 5)
dog2 = PetDog("Rocky",4, 15)
dog3 = PetDog("Fido", 6, 10)
names= [dog1, dog2, dog3]

dog1.play(dog2, dog3)
dog1.train()
dog1.do_a_trick()
```

```python
!pycodestyle my_script.py



class Family:
  def __init__(self,last_name, members):
    self.last_name = last_name
    self.members = members

  def born(self, **kwargs):
    self.members.append(kwargs)
    print(f"Congragulations {self.last_name} family on your new baby, {kwargs.get('name')}!")

  def is_18(self, name):
    for member in self.members:
      if member['name'] == name:
        return member['age'] > 18
    return True

  def family_presentation(self):
    print(f'family:{self.last_name}')
    for member in self.members:
      print(member)

f1 = Family("Fernandez",  [{'name':'Michael','age':35,'gender':'Male','is_child':False}, {'name':'Sarah','age':32,'gender':'Female','is_child':False}])
f1.born(name='Luna', age=0, gender='Female', is_child=True)
print(f1.is_18("Sarah"))
f1.family_presentation()

```
