1. Recap – What is a float? What is the difference between an integer and a float?

```python
#1 An integer is a whole number (1,2,3 or -1,2,3) and a float contains a decimal point (1.5, 2.0, 3.8)
```

2. Create a list containing the following sequence of floats and integers (it should be a list with mixed types): 1.5, 2, 2.5, 3, 3.5, 4, 4.5, 5 (don’t hard-code the sequence).

```python
list_start = 1.5
mixed_list =[]
while list_start <= 5:
  mixed_list.append(list_start)
  list_start += 0.5
print(mixed_list)
```

3. Can you think of another way to generate a sequence of floats?

```python
#list comprehension
  # Starts at 1.5 and loops 8 times (range(8)), incrementing by 0.5 each time.
  # i starts at 0, so the first value is 1.5 + (0 * 0.5) = 1.5.
  # Each iteration increases i by 1, multiplying by 0.5 to generate the next float.
  # The loop stops after 8 iterations, ending at 5.0.
another_example =[1.5 + i * 0.5 for i in range (8)]
print(another_example)
```

```python
ticketPrice = [0, 10, 15]
age=[]
```

2. Ask a family the age of each person who wants a ticket.

```python
ticket_prices = [0, 10, 15]
ages = []

while True:
    age = input("Enter age (or type 'Quit' to stop): ")

    if age == "Quit":
        break

    age = int(age)

    if age < 3:
        print(f"Ticket is ${ticket_prices[0]}")
    elif age <= 12:
        print(f"Ticket is ${ticket_prices[1]}")
    else:
        print(f"Ticket is ${ticket_prices[2]}")

    ages.append(age)
print("Ages collected:", ages)

```

3. Store the total cost of all the family’s tickets and print it out.

```python
total_ticket_price = 0
for age in ages:
   if age < 3:
    total_ticket_price += ticket_prices[0]
    print(f"Ticket is ${ticket_prices[0]}")
   elif age <= 12:
    total_ticket_price += ticket_prices[1]
    print(f"Ticket is ${ticket_prices[1]}")
   else:
    total_ticket_price += ticket_prices[2]
    print(f"Ticket is ${ticket_prices[2]}")

print(f'${total_ticket_price: .2f}')
```
