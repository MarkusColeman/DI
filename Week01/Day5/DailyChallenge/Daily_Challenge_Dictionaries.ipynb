 * Make sure the indexes are stored in a list and those lists are values.

Examples

"dodo" ➞ { "d": [0, 2], "o": [1, 3] }

"froggy" ➞ { "f":  [0], "r": [1], "o": [2], "g": [3, 4], "y": [5] }

"grapes" ➞ { "g": [0], "r": [1], "a": [2], "p": [3]}





```python
userWord = input("Enter a Word : ")

wordDict = {}

for index, letter in enumerate(userWord):
  key = (letter)
  value = (index)

  if letter in wordDict:
     wordDict[letter].append(index)
  else:
      wordDict[letter] = [index]
print(wordDict)
```

```python
purchaseItems = {"Jeans":"85",
                 "Shoes":"130",
                 "Jacket":"60",
                 "Baseball Cap":"30",
                 "Shirt":"50",
                 "Living room table":"900",
                 "Living room Chairs":"400"

}

walletVal = "$1,000"
walletNum = walletVal.replace("$", "").replace(",", "")
wallet = int(walletNum)

affordable = []

for product, price in purchaseItems.items():
  key = (product)
  value = int(price)

  if value <= wallet:
    affordable.append(product)

affordable.sort()

if not affordable:
    print("Nothing")
else:
    print(affordable)
```
